## Task 1

Create this data:

```
Months = Jan, Feb, Mar, Apr, May
Sales  = 20, 35, 30, 50, 45
```

Create a **line chart**.

Add:

- Title
- X-axis label
- Y-axis label

````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 30, 50, 45]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
````

## Task 2

Create a bar chart for:

```
Python      40
AI          55
Web         35
Design      25
```

````python
import matplotlib.pyplot as plt

courses = ["Python", "AI", "Web", "Design"]
students = [40, 55, 35, 25]

plt.bar(courses, students)

plt.title("Students by Course")
plt.xlabel("Courses")
plt.ylabel("Number of Students")

plt.show()
````

## Task 3

Create a scatter plot using:

```
Study Hours = 1, 2, 3, 4, 5, 6
Marks       = 40, 50, 60, 70, 82, 90
```

````python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
marks = [40, 50, 60, 70, 82, 90]

plt.scatter(study_hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
````

## Task 4

Create a histogram using these marks:

```
45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90
```

````python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90]

plt.hist(marks, bins=5)

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
````

## Task 5

Create a pie chart for:

```
Food       40
Transport  20
Shopping   25
Other      15
```

Display percentages.

````python
import matplotlib.pyplot as plt

categories = ["Food", "Transport", "Shopping", "Other"]
expenses = [40, 20, 25, 15]

plt.pie(expenses, labels=categories, autopct="%1.1f%%")

plt.title("Expenses Distribution")

plt.show()
````
